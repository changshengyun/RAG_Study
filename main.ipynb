```python
# 分片
from typing import List

def split_into_chunks(doc_file: str) -> List[str]:
    with open(doc_file, 'r',encoding="utf-8") as file:
        content = file.read()
    return [chunk for chunk in content.split("\n\n")]

chunks = split_into_chunks("doc.md") 

for i, chunk in enumerate(chunks):
    print(f"[{i}] {chunk}\n")
```

```python
# 索引
from sentence_transformers import SentenceTransformer
from typing import List

embedding_model = SentenceTransformer("shibing624/text2vec-base-chinese")

def embed_chunk(chunk: str) -> List[float]:
    embedding = embedding_model.encode(chunk)
    return embedding.tolist()

test_embedding = embed_chunk("测试内容")
print(len(test_embedding))
print(test_embedding)
```

```python
# 给所有片段生成对应向量
embeddings = [embed_chunk(chunk) for chunk in chunks]
print(len(embeddings))
print(embeddings[0])
```

```python
# 存入chromadb数据库
import chromadb

# 内存型 EphemeralClient
chromadb_client = chromadb.EphemeralClient()
# 磁盘型 PersistentClient
# chromadb_cient = chromadb.PersistentClient()

 # 获取/创建名为 default 的向量集合（类似数据表）
chromadb_collection = chromadb_client.get_or_create_collection(name="default")

# 定义函数：将文本块+向量存入向量库
def save_embeddings(chunks: List[str], embeddings: List[List[float]]) -> None:
    # 自动生成唯一字符串ID：0、1、2...
    ids = [str(i) for i in range(len(chunks))]
    # 批量插入向量数据
    chromadb_collection.add(
        documents=chunks,
        embeddings=embeddings,
        ids=ids
    )

# 调用函数，把分块文本和向量写入向量库
save_embeddings(chunks, embeddings)
```

```python
# retrieve 召回
from typing import List

def retrieve(query: str, top_k: int) -> List[str]:
    query_embedding = embed_chunk(query)
    results = chromadb_collection.query(
        query_embeddings=[query_embedding],
        n_results=top_k
    )
    return results['documents'][0]

query = "韩立的三叔都做了什么？"
retrieved_chunks = retrieve(query, 10)

for i, chunk in enumerate(retrieved_chunks):
    print(f"[{i}] {chunk}\n")
```

```python
# rerank 重排

from sentence_transformers import CrossEncoder
from typing import List

def rerank(query: str, retrieved_chunks: List[str], top_k: int) -> List[str]:
    cross_encoder = CrossEncoder('cross-encoder/mmarco-mMiniLMv2-L12-H384-v1')
    pairs = [(query, chunk) for chunk in retrieved_chunks]
    scores = cross_encoder.predict(pairs)

    chunk_with_score_list = [(chunk, score) for chunk, score in zip(retrieved_chunks, scores)]
    chunk_with_score_list.sort(key=lambda pair: pair[1], reverse=True)

    return [chunk for chunk, _ in chunk_with_score_list][:top_k]

reranked_chunks = rerank(query, retrieved_chunks, 3)
```

```python
from dotenv import load_dotenv
# from google import genai
from openai import OpenAI
# 加载.env环境变量文件
import os

load_dotenv()
# 自动读取.env中的GEMINI_API_KEY初始化Gemini客户端
# google_client = genai.Client()

# 初始化DeepSeek客户端（替代原来的google_client）
ds_client = OpenAI(
    api_key=os.getenv("DEEPSEEK_API_KEY"),
    base_url="https://api.deepseek.com/v1"
)

def generate(query: str, chunks: List[str]) -> str:
    chunk_text = "\n\n".join(chunks)
    # 构建RAG提示词，逻辑和你原来完全一致
    system_prompt = "你是一位知识助手，请根据用户的问题和提供的参考片段生成准确的回答，绝对不能编造片段以外的信息。"
    user_content = f"""
用户问题：{query}
相关参考片段：
{chunk_text}
"""
    messages = [
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": user_content}
    ]
    # 调用DeepSeek对话模型
    response = ds_client.chat.completions.create(
        model="deepseek-chat", # 通用对话模型；需要推理换成 deepseek-reasoner
        messages=messages,
        temperature=0.1 # 低温度，减少幻觉，适合RAG问答
    )
    # 返回模型回答文本
    return response.choices[0].message.content
    
# 调用生成函数，传入用户提问与重排后的高相关文本片段
answer = generate(query, reranked_chunks)
# 打印最终大模型回答
print(answer)
```
